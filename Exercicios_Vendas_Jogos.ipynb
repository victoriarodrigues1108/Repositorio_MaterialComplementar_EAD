**Exercício 7:**
Qual é a **média das Vendas Globais** de todos os jogos? E qual é a média dos anos de lançamento?

*Dica: Use o comando `.mean()` na coluna desejada.*

```python
# Seu código aqui
media_vendas = df['Vendas Globais'].mean()
print(f'A média de vendas é: {media_vendas}')
ano_lancamento = df.groupby(by='Ano')['Vendas Globais'].mean()
print(f'A média dos anos de lançamento é: {ano_lancamento}')
```
